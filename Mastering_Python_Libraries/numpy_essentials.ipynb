### Question 3:

Consider the flower petal $r(\theta) = 1 + \frac{3}{4}\sin(3\theta)$ for $0 \le \theta \le 2\pi$

1. Make a plot of the flower (find x and y points)

```python
theta = np.linspace(0, 2 * np.pi, 1_000)
r = 1 + (3 / 4) * np.sin(3 * theta)
x = r * np.cos(theta)
y = r * np.sin(theta)
plt.plot(x, y)
```

2. Compute the area using the calculus formula $A = \int_0^{2\pi}\frac{1}{2} r^2 d\theta $

```python
A = 1/2 * sum(r**2) * 
A(theta[1]-theta[0])
```

### Question 4:

The power emitted by a blackbody is $P = A\sigma \epsilon T^4$. After measuring the temperature of a star you find that 
$T(t) = T_0 \frac{1}{1 +e^{-kt}}$. Plot the total energy emitted by the star as a function of time using the fact that 
 $E(t) = \int_{t'=0}^{t'=t} P(t') dt' $

```python
kt = np.linspace(0, 3, 100)
P = (1 / (1 + np.exp(-kt))) ** 4
E = np.cumsum(P) * (kt[1] - kt[0])
```

```python
plt.plot(kt, E)
plt.xlabel("$kt$", fontsize=20)
plt.ylabel(r"$\left(\frac{k}{A \sigma \epsilon T_0^4}\right) E(kt)$", fontsize=20)
plt.show()
```

### Question 1

Let $f(x,y) = e^{-(x^2+y^2)} \cdot \sin(x) \text{ for } -2 \le x \le 2 \text{ and } -2 \le y \le 2$

1. Make a contour plot of f

```python
x = np.linspace(-2, 2, 1_000)
y = np.linspace(-2, 2, 1_000)

xv, yv = np.meshgrid(x, y)
f = np.exp(-(xv**2 + yv**2)) * np.sin(xv)

plt.contourf(xv, yv, f, levels=100)
plt.colorbar()
```

2. Find the volume $|f(x,y)|$ in the specified x and y range 

```python
sum(np.abs(f.ravel())) * np.diff(x)[0] * np.diff(y)[0]
```

3. Find the volume  $|f(x,y)|$ only in the region where $\sqrt{x^2+y^2} > 0.5$

```python
sum(np.abs(f[xv**2 + yv**2 > 0.5**2].ravel())) * np.diff(x)[0] * np.diff(y)[0]
```

An electric field is given by $\vec{E}(z,t) = E_0 \cos (z-t) \hat{x} + 2E_0 \cos(z-t+\pi/2) \hat{y}$. 

1. Find the magnetic field for all $z \in [0,4\pi]$ and $t \in [0,10]$ using $\vec{B}(z,t) = \hat{z} \times \vec{E}(z,t)$

```python
z = np.linspace(0, 4 * np.pi, 100)
t = np.linspace(0, 10, 100)

tv, zv = np.meshgrid(t, z)

Ex = np.cos(zv - tv)
Ey = 2 * np.cos(zv - tv + np.pi / 2)
Ez = 0 * tv
```

```python
E = np.array([Ex, Ey, Ez])
E = np.swapaxes(E, 0, -1)
```

```python
B = np.cross(np.array([0, 0, 1]), E)
```
